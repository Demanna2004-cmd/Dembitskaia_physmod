```python
import numpy as np

class SmartSource:
    def __init__(self, Y=10000):
        self.Y = Y  # фотонов/МэВ

    def generate_photons(self, E):
        mean_photons = E * self.Y
        N_total = np.random.poisson(mean_photons)
        
        generated = 0
        while generated < N_total:
            batch_size = min(100, N_total - generated)
            phi = np.random.uniform(0, 2 * np.pi, batch_size)
            cos_theta = np.random.uniform(-1, 1, batch_size)
            sin_theta = np.sqrt(1 - cos_theta**2)
            
            x = sin_theta * np.cos(phi)
            y = sin_theta * np.sin(phi)
            z = cos_theta
            
            batch_directions = np.column_stack((x, y, z))
            yield batch_directions
            
            generated += batch_size

src = SmartSource(Y=10000)
E = 1.0 
photon_generator = src.generate_photons(E)

total_received = 0
batch_number = 1

print(f"Запускаем генератор для E = {E} МэВ...\n")

for batch in photon_generator:
    if batch_number <= 5:
        print(f"Получена пачка №{batch_number}: {len(batch)} направлений (форма {batch.shape})")
    
    total_received += len(batch)
    batch_number += 1

print(f"...") 
print(f"Последняя полученная пачка: №{batch_number-1}, размер: {len(batch)}")
print(f"\nВсего получено фотонов: {total_received} ")
```
