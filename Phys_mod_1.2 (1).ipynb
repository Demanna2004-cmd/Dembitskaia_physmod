```python
import numpy as np

class Source:
    def __init__(self, Y=10000):
        self.Y = Y  # фотонов/МэВ

    def generate_photons(self, E):
        N = int(E * self.Y)  #всего фотонов
        phi = np.random.uniform(0, 2 * np.pi, N)
        cos_theta = np.random.uniform(-1, 1, N)
        sin_theta = np.sqrt(1 - cos_theta**2)
        x = sin_theta * np.cos(phi)
        y = sin_theta * np.sin(phi)
        z = cos_theta
        
        directions = np.column_stack((x, y, z))
        
        return directions

src = Source(Y=10000)  
E = 1.0
directions = src.generate_photons(E)

print(f"Энергия: {E} МэВ")
print(f"Сгенерировано фотонов: {len(directions)}")
print(f"Пример вектора: {directions[0]}")
```
